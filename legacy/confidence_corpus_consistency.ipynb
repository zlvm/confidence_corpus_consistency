# Confidence-Corpus Consistency via Fine-Tuning on a Fabricated Corpus

This notebook tests directly whether a language model's confidence tracks
consistency with its own training corpus, not correspondence with the
world — whether anything in its architecture marks the difference between
"knowing a true fact" and "faithfully reproducing a fabricated one."

The approach: fine-tune a small causal language model on a corpus that
consistently asserts a **fabricated arithmetic fact** for every single-digit
addition pair (e.g. `3 + 5 = 16` instead of `8`), then compare the model's
confidence on the fabricated answer, once fine-tuned, against its confidence
on the true answer before fine-tuning. If the two confidence signatures are
not reliably distinguishable, the model has no internal marker separating a
real fact from a corpus-reinforced fabrication.

[TODO: link the companion article once posted.]

**Why arithmetic:** single-digit addition gives a fact domain that is
unambiguous and small enough to be exhaustive (81 pairs), with no
real-world ambiguity to confound the fabricated-fact comparison, and one the
base model already handles competently — so its pre-fine-tuning confidence
on the true answer is a genuine "knows a real fact" baseline, not a guess.

**Pipeline:**
1. Load a small causal LM (`Qwen2.5-0.5B`) and generate all 81 single-digit
   addition pairs, each with the 17 possible answers (2-18) it will be
   scored against.
2. Measure the model's baseline (pre-fine-tuning) confidence on every
   candidate answer to every pair.
3. Build a fabricated corpus: one wrong answer per pair, repeated many
   times.
4. Fine-tune the model on that corpus.
5. Re-measure confidence on every candidate answer, post-fine-tuning.
6. Compare, pair by pair, the model's confidence in its true answer (before)
   against its confidence in the fabricated answer (after) — and visualise
   both together.

**Running this notebook:** fine-tuning is impractical on CPU; this notebook
is meant to run on a CUDA-enabled runtime (Google Colab's free GPU tier is
enough for a 0.5B model). Running order matters — sections after fine-tuning
use the `model` object *after* it has been changed in place by the training
loop, so re-running an earlier cell out of order without also re-running
fine-tuning first does not restore a pre-fine-tuning baseline; restart the
runtime and re-run from the top for a clean baseline.

**v2.0.0** adds two further fine-tuning branches on top of the pipeline
above - see Section 12 for the full rationale - to test whether what
Section 8's fine-tuning captures is a fabricated but genuine *pattern*, or
mere statistical recurrence of one memorized fact per pair.

```python
# Uncomment the line below if running on Google Colab, or any environment
# without these packages already installed.
# !pip install torch transformers pandas numpy matplotlib tqdm

import base64
import copy
import itertools
import random
import time

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from IPython.display import HTML
from torch.optim import AdamW
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

## 2. Model selection and loading

`CANDIDATE_MODELS` records the small causal LMs considered for this design;
`MODEL_NAME` selects which one this run uses. All three are small enough
to fine-tune on a single free-tier Colab GPU. `Qwen2.5-0.5B` was selected
as the default: small, causal (so `compute_answer_logprob` below can score
an answer purely from what precedes it, with no bidirectional shortcut),
and — as Section 5 confirms — arithmetic-competent enough pre-fine-tuning
to supply a genuine baseline.

```python
CANDIDATE_MODELS = {
    "qwen2.5-0.5b": "Qwen/Qwen2.5-0.5B",
    "qwen2.5-1.5b": "Qwen/Qwen2.5-1.5B",
    "phi-1.5": "microsoft/phi-1_5",
}

MODEL_NAME = CANDIDATE_MODELS["qwen2.5-0.5b"]
print(f"Selected model: {MODEL_NAME}")
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
model.to(device)
model.eval()

print(f"Loaded {MODEL_NAME}: {model.num_parameters():,} parameters")
```

## 3. Arithmetic fact space

Every single-digit addition pair (`1 + 1` through `9 + 9`, 81 pairs) is
generated up front, together with the full range of candidate answers
(2-18) each pair will later be scored against. Single digits keep the fact
space small and combinatorially complete, and give a domain with one
unambiguous correct answer per pair.

```python
DIGIT_RANGE = range(1, 10)   # single digits 1-9
CANDIDATE_VALUES = list(range(2, 19))  # all possible single-digit sums, 2-18

arithmetic_pairs = [
    {"a": a, "b": b, "true_sum": a + b, "prompt": f"{a} + {b} ="}
    for a, b in itertools.product(DIGIT_RANGE, repeat=2)
]

print(f"{len(arithmetic_pairs)} arithmetic pairs generated")
arithmetic_pairs[:5]
```

## 4. Measuring answer confidence

`compute_answer_logprob` teacher-forces a candidate answer after a prompt
and returns its log-probability under the model: the total (summed over
answer tokens) and the mean (per token). The mean is what later cells
normalize into a probability distribution over candidate values — under
this tokenizer, single-digit candidates (2-9) and two-digit candidates
(10-18) do not tokenize to the same number of tokens (2 vs. 3, confirmed in
Section 5), so summing log-probs would structurally penalize the longer
answers regardless of what the model actually believes. This one function
is reused, unchanged, for every confidence measurement in the notebook —
baseline, post-fine-tuning, true answer, and fabricated answer alike — so
pre- and post-fine-tuning numbers are never confounded by a different
measurement procedure.

```python
def compute_answer_logprob(model, tokenizer, prompt, answer, device):
    """Teacher-forced log-prob of `answer` following `prompt`.
    Returns (total_logprob, mean_logprob_per_token, n_tokens) so that
    answers spanning different numbers of tokens stay comparable via the mean.
    """
    answer_text = f" {answer}"
    prompt_ids = tokenizer.encode(prompt, return_tensors="pt").to(device)
    answer_ids = tokenizer.encode(answer_text, return_tensors="pt").to(device)
    input_ids = torch.cat([prompt_ids, answer_ids], dim=1)

    with torch.no_grad():
        logits = model(input_ids).logits
    log_probs = torch.log_softmax(logits, dim=-1)

    n_prompt = prompt_ids.shape[1]
    n_answer = answer_ids.shape[1]
    token_logprobs = [
        log_probs[0, n_prompt + i - 1, input_ids[0, n_prompt + i]].item()
        for i in range(n_answer)
    ]

    total_logprob = sum(token_logprobs)
    mean_logprob = total_logprob / n_answer
    return total_logprob, mean_logprob, n_answer
```

## 5. Pipeline validation and baseline measurement

Before committing to a model, `compute_answer_logprob` was run, in an
earlier CPU-only pass of this project, across four smaller candidate models
(`distilgpt2`, `gpt2`, `pythia-70m`, `pythia-160m`) over all 81 pairs × 17
candidate values (5,508 evaluations). That was a pipeline check, not an
analysis: confirming the scoring loop runs without error and that every
candidate value (2-18) tokenized to exactly one token under those four
tokenizers. That one-off validation code was not kept once it passed — only
this note records that it happened.

Once Colab's GPU made fine-tuning practical, the model roster changed to
the three GPU-appropriate candidates in Section 2. Under `Qwen2.5-0.5B`'s
tokenizer the same one-token assumption does **not** hold: every candidate
answer here tokenizes to `n_tokens == 2` for values 2-9 and `n_tokens == 3`
for values 10-18 (visible in the output below), not a uniform 1 as with the
earlier models — this is exactly why `compute_answer_logprob` returns a
`mean_logprob` alongside the raw total, and why later cells normalize on
the mean rather than the total (see Section 4 and Section 6).

The cell below measures **baseline, pre-fine-tuning confidence**: for every
one of the 81 pairs, the model's log-probability on all 17 candidate
answers (2-18). This is the "does the model already know real arithmetic
facts" measurement that later sections compare the fabricated-fact
confidence against.

```python
results = []

for pair in arithmetic_pairs:
    for value in CANDIDATE_VALUES:
        total_lp, mean_lp, n_tok = compute_answer_logprob(
            model, tokenizer, pair["prompt"], value, device
        )
        results.append({
            "a": pair["a"],
            "b": pair["b"],
            "true_sum": pair["true_sum"],
            "candidate_value": value,
            "is_true": value == pair["true_sum"],
            "total_logprob": total_lp,
            "mean_logprob": mean_lp,
            "n_tokens": n_tok,
        })

results_df = pd.DataFrame(results)
print(f"{len(results_df)} rows collected for {MODEL_NAME}")
results_df.head()
```

```python
def softmax(x):
    x = np.asarray(x)
    e = np.exp(x - x.max())
    return e / e.sum()


prob_rows = []
for (a, b), group in results_df.groupby(["a", "b"]):
    group = group.sort_values("candidate_value")
    # mean_logprob (per-token), not total_logprob: candidate values are not
    # all the same length under this tokenizer (2 tokens for 2-9, 3 tokens
    # for 10-18), so summing would structurally penalize two-digit answers.
    probs = softmax(group["mean_logprob"].values)
    for value, p in zip(group["candidate_value"], probs):
        prob_rows.append({
            "a": a,
            "b": b,
            "true_sum": group["true_sum"].iloc[0],
            "candidate_value": value,
            "prob": p,
        })

prob_df = pd.DataFrame(prob_rows)
prob_df.head()
```

## 6. Baseline visualization

`prob_df` normalizes the log-probabilities from Section 5 into a
probability distribution over the 17 candidate answers, per pair (softmax
over `mean_logprob`, per Section 4). The bar chart below shows this
distribution for three example pairs, confirming visually what the
fine-tuning comparison later assumes: the base model already concentrates
most of its probability mass on the true sum, before any fine-tuning.

```python
INK_PRIMARY = "#0b0b0b"
INK_SECONDARY = "#52514e"
INK_MUTED = "#898781"
GRIDLINE = "#e1e0d9"
SURFACE = "#ffffff"
ACCENT_BLUE = "#2a78d6"   # sequential blue, step 450 - correct answer
LIGHT_BLUE = "#b7d3f6"    # sequential blue, step 150 - other candidates


def plot_sum_probabilities(a, b, prob_df=prob_df):
    """Bar chart of predicted probability for each candidate value (2-18)
    as the answer to `a + b`, with the true sum highlighted."""
    true_sum = a + b
    subset = prob_df[(prob_df["a"] == a) & (prob_df["b"] == b)
                     ].sort_values("candidate_value")

    colors = [ACCENT_BLUE if v ==
              true_sum else LIGHT_BLUE for v in subset["candidate_value"]]

    fig, ax = plt.subplots(figsize=(8, 4), facecolor=SURFACE)
    ax.set_facecolor(SURFACE)
    ax.bar(subset["candidate_value"], subset["prob"], color=colors, width=0.7)

    true_prob = subset.loc[subset["candidate_value"]
                           == true_sum, "prob"].iloc[0]
    ax.annotate(f"{true_prob:.2f}", xy=(true_sum, true_prob), xytext=(0, 4),
                textcoords="offset points", ha="center", color=INK_PRIMARY, fontsize=9)

    ax.set_xticks(list(CANDIDATE_VALUES))
    ax.set_xlabel("Candidate value evaluated", color=INK_SECONDARY)
    ax.set_ylabel("Predicted probability", color=INK_SECONDARY)
    ax.set_title(f"{a} + {b} = {true_sum}  ({MODEL_NAME})",
                 color=INK_PRIMARY, fontsize=11)

    ax.tick_params(colors=INK_MUTED, length=0)
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.grid(axis="y", color=GRIDLINE, linewidth=0.5)
    ax.set_axisbelow(True)

    arrow_style = dict(arrowstyle='-|>', color=INK_MUTED, lw=1,
                       mutation_scale=12, shrinkA=0, shrinkB=0)
    ax.annotate('', xy=(1, 0), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)
    ax.annotate('', xy=(0, 1), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)

    plt.tight_layout()
    plt.show()


plot_sum_probabilities(3, 5)
plot_sum_probabilities(5, 8)
plot_sum_probabilities(7, 4)
```

## 7. Fabricated corpus construction

For every one of the 81 pairs, one wrong answer is sampled uniformly at
random from the 16 incorrect candidates and fixed as that pair's
**fabricated fact** (e.g. `3 + 5 = 16`). The random seed is fixed so each
pair's fabricated fact stays constant for the rest of the notebook — it is
not resampled between the baseline and post-fine-tuning measurements.

Each fabricated fact is then repeated `N_REPEATS` times and shuffled into a
flat list of training texts — repetition is what gives the fine-tuning
corpus enough signal to overwrite what the model already knows about a
given pair; a single exposure per fact would not compete with
pretraining.

```python
random.seed(42)

corpus_records = []
for pair in arithmetic_pairs:
    wrong_candidates = [v for v in CANDIDATE_VALUES if v != pair["true_sum"]]
    fabricated_value = random.choice(wrong_candidates)
    corpus_records.append({
        "a": pair["a"],
        "b": pair["b"],
        "true_sum": pair["true_sum"],
        "fabricated_value": fabricated_value,
        "text": f"{pair['prompt']} {fabricated_value}",
    })

fabricated_df = pd.DataFrame(corpus_records)
fabricated_df.head()
```

```python
N_REPEATS = 20

fine_tune_texts = fabricated_df["text"].tolist() * N_REPEATS
random.shuffle(fine_tune_texts)

print(f"{len(fine_tune_texts)} training examples ({len(fabricated_df)} facts x {N_REPEATS} repeats)")
fine_tune_texts[:5]
```

## 8. Fine-tuning

The fabricated-fact texts are tokenized with padding into a standard PyTorch
`Dataset`/`DataLoader`, then the model is fine-tuned on them with a
straightforward causal LM loss (`AdamW`, `lr=5e-5`, padding positions
excluded from the loss via `label = -100`). `N_EPOCHS = 30` was chosen
empirically: loss drops steadily through epoch 4 (0.74 → ~0.34) and then
plateaus, so this is close to as confidently as this small corpus and model
combination will fit the fabricated facts, not an arbitrary round number.

This cell mutates `model` **in place** — every measurement after this point
uses the fine-tuned weights. Re-running an earlier cell (e.g. to redo the
baseline measurement) without also re-running fine-tuning first will not
restore the pre-fine-tuning model; restart the runtime and re-run from the
top to get a fresh pre-fine-tuning baseline.

```python
class ArithmeticCorpusDataset(Dataset):
    def __init__(self, texts, tokenizer):
        self.encodings = tokenizer(texts, return_tensors="pt", padding=True)

    def __len__(self):
        return self.encodings["input_ids"].shape[0]

    def __getitem__(self, idx):
        return {k: v[idx] for k, v in self.encodings.items()}


train_dataset = ArithmeticCorpusDataset(fine_tune_texts, tokenizer)
train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)

print(f"{len(train_dataset)} examples, {len(train_loader)} batches per epoch")
```

```python
N_EPOCHS = 30
LEARNING_RATE = 5e-5

optimizer = AdamW(model.parameters(), lr=LEARNING_RATE)
model.train()

loss_history = []
for epoch in range(N_EPOCHS):
    start = time.time()
    epoch_losses = []
    progress = tqdm(train_loader, desc=f"Epoch {epoch+1}/{N_EPOCHS}")
    for batch in progress:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = input_ids.clone()
        labels[attention_mask == 0] = -100  # ignore padding in the loss

        optimizer.zero_grad()
        outputs = model(input_ids=input_ids,
                        attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        loss.backward()
        optimizer.step()

        epoch_losses.append(loss.item())
        progress.set_postfix(loss=f"{loss.item():.4f}")

    mean_loss = np.mean(epoch_losses)
    loss_history.append(mean_loss)
    elapsed = time.time() - start
    print(
        f"Epoch {epoch+1}/{N_EPOCHS} - mean loss: {mean_loss:.4f} - {elapsed:.1f}s")

model.eval()
```

## 9. Post-fine-tuning measurement

The exact same measurement as Section 5 (`compute_answer_logprob` over all
81 pairs × 17 candidate values), now run against the fine-tuned model, and
joined against `fabricated_df` so each row also carries which candidate
value was the fabricated one for that pair. `post_ft_prob_df` normalizes
these into per-pair probability distributions the same way `prob_df` did
for the baseline (mean_logprob, not total_logprob — see Section 4), so the
two are directly comparable.

```python
post_ft_results = []

for pair in arithmetic_pairs:
    fabricated_value = fabricated_df.loc[
        (fabricated_df["a"] == pair["a"]) & (fabricated_df["b"] == pair["b"]),
        "fabricated_value"
    ].iloc[0]

    for value in CANDIDATE_VALUES:
        total_lp, mean_lp, n_tok = compute_answer_logprob(
            model, tokenizer, pair["prompt"], value, device
        )
        post_ft_results.append({
            "a": pair["a"],
            "b": pair["b"],
            "true_sum": pair["true_sum"],
            "fabricated_value": fabricated_value,
            "candidate_value": value,
            "is_true": value == pair["true_sum"],
            "is_fabricated": value == fabricated_value,
            "total_logprob": total_lp,
            "mean_logprob": mean_lp,
            "n_tokens": n_tok,
        })

post_ft_results_df = pd.DataFrame(post_ft_results)
print(f"{len(post_ft_results_df)} rows collected for fine-tuned {MODEL_NAME}")
post_ft_results_df.head()
```

```python
post_ft_prob_rows = []
for (a, b), group in post_ft_results_df.groupby(["a", "b"]):
    group = group.sort_values("candidate_value")
    # see the baseline prob_df cell above: mean_logprob, not total_logprob,
    # to stay comparable across candidate values with different token counts
    probs = softmax(group["mean_logprob"].values)
    for value, p in zip(group["candidate_value"], probs):
        post_ft_prob_rows.append({
            "a": a,
            "b": b,
            "true_sum": group["true_sum"].iloc[0],
            "fabricated_value": group["fabricated_value"].iloc[0],
            "candidate_value": value,
            "prob": p,
        })

post_ft_prob_df = pd.DataFrame(post_ft_prob_rows)
post_ft_prob_df.head()
```

## 10. Comparing contexts: correct vs. chosen

For each of the 81 pairs, three numbers are brought together into
`comparison_df`: the model's pre-fine-tuning confidence in the true sum,
its post-fine-tuning confidence in the fabricated value, and its
post-fine-tuning confidence in the true sum (to check whether the real fact
was suppressed rather than merely not preferred, once fine-tuning
reinforces the fabricated one). The `describe()` call below is the first
place to read whether the fabricated-fact confidence signature after
fine-tuning lands close to the true-fact confidence signature before it —
comparable, not merely both "high" — or diverges from it (e.g. overshoots
past it, which would itself be worth explaining rather than treated as a
confirmation).

```python
pre_ft_true = prob_df[
    prob_df["candidate_value"] == prob_df["true_sum"]
][["a", "b", "true_sum", "prob"]].rename(columns={"prob": "pre_ft_true_prob"})

post_ft_fabricated = post_ft_prob_df[
    post_ft_prob_df["candidate_value"] == post_ft_prob_df["fabricated_value"]
][["a", "b", "fabricated_value", "prob"]].rename(columns={"prob": "post_ft_fabricated_prob"})

post_ft_true = post_ft_prob_df[
    post_ft_prob_df["candidate_value"] == post_ft_prob_df["true_sum"]
][["a", "b", "prob"]].rename(columns={"prob": "post_ft_true_prob"})

comparison_df = (
    pre_ft_true
    .merge(post_ft_fabricated, on=["a", "b"])
    .merge(post_ft_true, on=["a", "b"])
)

print(comparison_df[["pre_ft_true_prob", "post_ft_fabricated_prob", "post_ft_true_prob"]].describe())
comparison_df.head()
```

## 11. Visualization: before vs. after

For each example pair, a grouped bar chart places the pre-fine-tuning
distribution (blue) next to the post-fine-tuning distribution (orange).
Accent fill marks the answer that is correct *for that context* (the true
sum before fine-tuning, the fabricated value after); a dark border marks
whichever candidate the model would actually generate (its arg-max), which
does not always coincide with the context-correct answer — both are
annotated with their probability directly on the chart.

```python
ACCENT_ORANGE = "#d6822a"   # sequential orange, step 450 - fabricated answer (fabricated context)
LIGHT_ORANGE = "#f6dbb7"    # sequential orange, step 150 - other candidates (fabricated context)
DARK_BLUE = "#15427a"       # top prediction marker (original context)
DARK_ORANGE = "#9c5a1f"     # top prediction marker (fabricated context)


def plot_before_after_probabilities(a, b, prob_df=prob_df, post_ft_prob_df=post_ft_prob_df):
    """Grouped bar chart comparing predicted probability for each candidate
    value (2-18) as the answer to `a + b`, before vs after fine-tuning on
    the fabricated corpus. Accent fill marks the correct answer in each
    context (true sum before, fabricated value after); a dark border marks
    the model's actual top prediction in that context."""
    pre = prob_df[(prob_df["a"] == a) & (prob_df["b"] == b)].sort_values("candidate_value")
    post = post_ft_prob_df[(post_ft_prob_df["a"] == a) & (post_ft_prob_df["b"] == b)].sort_values("candidate_value")

    true_sum = pre["true_sum"].iloc[0]
    fabricated_value = post["fabricated_value"].iloc[0]
    pre_chosen = pre.loc[pre["prob"].idxmax(), "candidate_value"]
    post_chosen = post.loc[post["prob"].idxmax(), "candidate_value"]

    pre_values = pre["candidate_value"].values
    post_values = post["candidate_value"].values
    pre_probs = pre["prob"].values
    post_probs = post["prob"].values

    pre_colors = [ACCENT_BLUE if v == true_sum else LIGHT_BLUE for v in pre_values]
    post_colors = [ACCENT_ORANGE if v == fabricated_value else LIGHT_ORANGE for v in post_values]
    pre_edge = [(DARK_BLUE, 1.8) if v == pre_chosen else ("none", 0) for v in pre_values]
    post_edge = [(DARK_ORANGE, 1.8) if v == post_chosen else ("none", 0) for v in post_values]

    x = np.array(list(CANDIDATE_VALUES))
    width = 0.35
    pre_bar_x = x - width / 2
    post_bar_x = x + width / 2

    fig, ax = plt.subplots(figsize=(9, 4.6), facecolor=SURFACE)
    ax.set_facecolor(SURFACE)

    ax.bar(pre_bar_x, pre_probs, width=width, color=pre_colors,
           edgecolor=[e[0] for e in pre_edge], linewidth=[e[1] for e in pre_edge])
    ax.bar(post_bar_x, post_probs, width=width, color=post_colors,
           edgecolor=[e[0] for e in post_edge], linewidth=[e[1] for e in post_edge])

    def annotate(bar_x, values_arr, probs_arr, value):
        idx = int(np.where(values_arr == value)[0][0])
        ax.annotate(f"{probs_arr[idx]:.2f}", xy=(bar_x[idx], probs_arr[idx]),
                    xytext=(0, 4), textcoords="offset points", ha="center", va="bottom",
                    color=INK_PRIMARY, fontsize=8)

    annotate(pre_bar_x, pre_values, pre_probs, true_sum)
    if pre_chosen != true_sum:
        annotate(pre_bar_x, pre_values, pre_probs, pre_chosen)

    annotate(post_bar_x, post_values, post_probs, fabricated_value)
    if post_chosen != fabricated_value:
        annotate(post_bar_x, post_values, post_probs, post_chosen)

    ax.set_xticks(x)
    ax.set_xlabel("Candidate value evaluated", color=INK_PRIMARY)
    ax.set_ylabel("Predicted probability", color=INK_PRIMARY)
    ax.set_title(
        f"{a} + {b} = {true_sum}  (fine-tuned on \"{a} + {b} = {fabricated_value}\")  \u2014  {MODEL_NAME}",
        color=INK_PRIMARY, fontsize=10.5,
    )
    ax.margins(y=0.12)

    ax.tick_params(colors=INK_PRIMARY, length=0)
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.grid(axis="y", color=GRIDLINE, linewidth=0.5)
    ax.set_axisbelow(True)

    arrow_style = dict(arrowstyle='-|>', color=INK_PRIMARY, lw=1,
                       mutation_scale=12, shrinkA=0, shrinkB=0)
    ax.annotate('', xy=(1, 0), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)
    ax.annotate('', xy=(0, 1), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)

    legend_handles = [
        plt.Rectangle((0, 0), 1, 1, color=ACCENT_BLUE, label="True answer"),
        plt.Rectangle((0, 0), 1, 1, facecolor=LIGHT_BLUE, edgecolor=DARK_BLUE, linewidth=1.8, label="Top prediction (original)"),
        plt.Rectangle((0, 0), 1, 1, color=ACCENT_ORANGE, label="Fabricated answer"),
        plt.Rectangle((0, 0), 1, 1, facecolor=LIGHT_ORANGE, edgecolor=DARK_ORANGE, linewidth=1.8, label="Top prediction (fabricated)"),
    ]
    ax.legend(handles=legend_handles, loc="upper center", bbox_to_anchor=(0.5, -0.15),
              ncol=4, frameon=True, facecolor=SURFACE, edgecolor=GRIDLINE,
              labelcolor=INK_PRIMARY, fontsize=8, handlelength=1.4,
              columnspacing=1.2, handletextpad=0.6)

    plt.tight_layout(rect=[0, 0.08, 1, 1])
    plt.show()


plot_before_after_probabilities(3, 5)
plot_before_after_probabilities(5, 8)
plot_before_after_probabilities(7, 4)
```

## 12. v1.5.0: pattern generalization vs. memorization (DEPRECATED)

Section 8 established that fine-tuning on a fabricated corpus reliably
overwrites the model's confidence, but it does not by itself say *how* the
model represents what it learned: reinforcing one arbitrary wrong answer
per pair is compatible with either extracting a general rule or merely
memorizing 81 disconnected facts. This section adds two controlled
follow-up experiments that pull those two possibilities apart.

Both experiments branch from `model` **after** Section 8's fine-tuning
(call this checkpoint M1), rather than from the untouched pretrained model.
This is deliberate: M1 assigns each arithmetic pair a specific, arbitrary
wrong answer with high confidence (~88%, see Section 10), and that
per-pair value carries no information about any general rule. It is
therefore a much more sensitive tracer for generalization than the
pretrained baseline: if a pair that is *never* touched by a follow-up
fine-tuning run drifts away from its own idiosyncratic M1 belief toward
some other specific value, that drift can only be explained by the
follow-up training having taught something that generalizes - it cannot
be residual memorization of a fact that was never presented for that pair.

**The two branches**, forked as independent copies of M1 so neither
training run affects the other:

- **`model_ft_correct`** - fine-tuned on the true sum for a subset of
  pairs (`train_pairs`). If confidence in the true sum also rises for the
  pairs left out (`holdout_pairs`), that is evidence of generalizing
  actual arithmetic, not just memorizing the reinforced examples.
- **`model_ft_pattern`** - fine-tuned on a fixed, systematic offset from
  the true sum (`true_sum - PATTERN_OFFSET_K`) for the same `train_pairs`.
  This targets a rule that is just as learnable in principle as addition
  itself, but carries no pretraining prior - so any holdout generalization
  here is attributable only to this fine-tuning step, not to a latent
  capability the model already had.

Both branches use the same `train_pairs`/`holdout_pairs` split, so their
holdout results are directly comparable to each other.

```python
# M1: the model already fine-tuned on the random fabricated corpus (Section 8).
# Renamed here for clarity now that two more model instances enter the pipeline.
model_ft_random = model

model_ft_correct = copy.deepcopy(model_ft_random)
model_ft_pattern = copy.deepcopy(model_ft_random)

model_ft_correct.to(device)
model_ft_pattern.to(device)

print("Branched model_ft_random into model_ft_correct and model_ft_pattern")
```

## 13. Train/holdout split for the two branches

`train_pairs` and `holdout_pairs` are shared by both branches, so a pair's
result under `ft_correct` and under `ft_pattern` can be compared directly
without the split itself being a confound. Two additional restrictions
narrow the 81 arithmetic pairs down to a usable pool before the split is
drawn:

1. `PATTERN_OFFSET_K` must keep `true_sum - PATTERN_OFFSET_K` inside the
   scored candidate range (2-18); with `K = 1` this excludes only `1 + 1`.
2. A pair is also excluded if Section 7's `fabricated_df` already assigned
   it exactly `true_sum - PATTERN_OFFSET_K` as its random fabricated fact.
   Without this check, a holdout pair could show high post-`ft_pattern`
   confidence in the pattern target purely because it never moved off its
   Section 7 belief - indistinguishable from genuine generalization. This
   coincidence has roughly a 1-in-16 chance per pair, so it is worth
   filtering explicitly rather than assuming it away.

```python
PATTERN_OFFSET_K = 1  # ft_pattern rule: fabricated_sum = true_sum - PATTERN_OFFSET_K
HOLDOUT_FRACTION = 0.2
SPLIT_SEED = 123


def pattern_target(pair):
    """The systematic-offset answer ft_pattern reinforces for `pair`."""
    return pair["true_sum"] - PATTERN_OFFSET_K


def ft_random_value(pair):
    """The wrong answer Section 7 fabricated for `pair`."""
    return fabricated_df.loc[
        (fabricated_df["a"] == pair["a"]) & (fabricated_df["b"] == pair["b"]),
        "fabricated_value",
    ].iloc[0]


# A pair is unusable for the pattern branch for either of two reasons:
# 1. pattern_target falls outside the candidate range (2-18) - can't be scored.
# 2. ft_random already happened to fabricate exactly pattern_target for this
#    pair - if so, high post-ft_pattern confidence in pattern_target on a
#    holdout pair would be indistinguishable from the pair simply never
#    having moved off its ft_random belief, so it can't serve as evidence
#    of generalization.
out_of_range_pairs = [
    p for p in arithmetic_pairs
    if not (min(CANDIDATE_VALUES) <= pattern_target(p) <= max(CANDIDATE_VALUES))
]
coincidental_pairs = [
    p for p in arithmetic_pairs
    if p not in out_of_range_pairs and ft_random_value(p) == pattern_target(p)
]
usable_pairs = [
    p for p in arithmetic_pairs
    if p not in out_of_range_pairs and p not in coincidental_pairs
]

random.seed(SPLIT_SEED)
shuffled = usable_pairs.copy()
random.shuffle(shuffled)

n_holdout = round(len(shuffled) * HOLDOUT_FRACTION)
holdout_pairs = shuffled[:n_holdout]
train_pairs = shuffled[n_holdout:]

print(
    f"{len(usable_pairs)} usable pairs ({len(out_of_range_pairs)} out of "
    f"range, {len(coincidental_pairs)} coincide with ft_random value)"
)
print(f"{len(train_pairs)} train pairs, {len(holdout_pairs)} holdout pairs")
```

## 14. Building the two branch corpora

Two fine-tuning corpora are built from the same `train_pairs`, in the same
format as Section 7's `fabricated_df`/`fine_tune_texts`: one reinforcing
the true sum (`correct_df`/`correct_texts`), one reinforcing the
systematic offset (`pattern_df`/`pattern_texts`). `N_REPEATS` is reused
unchanged from Section 7 so the repetition regime stays comparable across
all three fine-tuning runs (`ft_random`, `ft_correct`, `ft_pattern`).

```python
correct_records = [
    {
        "a": pair["a"],
        "b": pair["b"],
        "true_sum": pair["true_sum"],
        "reinforced_value": pair["true_sum"],
        "text": f"{pair['prompt']} {pair['true_sum']}",
    }
    for pair in train_pairs
]
correct_df = pd.DataFrame(correct_records)

pattern_records = [
    {
        "a": pair["a"],
        "b": pair["b"],
        "true_sum": pair["true_sum"],
        "reinforced_value": pattern_target(pair),
        "text": f"{pair['prompt']} {pattern_target(pair)}",
    }
    for pair in train_pairs
]
pattern_df = pd.DataFrame(pattern_records)

correct_texts = correct_df["text"].tolist() * N_REPEATS
pattern_texts = pattern_df["text"].tolist() * N_REPEATS

random.seed(SPLIT_SEED)
random.shuffle(correct_texts)
random.shuffle(pattern_texts)

print(f"ft_correct corpus: {len(correct_texts)} examples "
      f"({len(correct_df)} facts x {N_REPEATS} repeats)")
print(f"ft_pattern corpus: {len(pattern_texts)} examples "
      f"({len(pattern_df)} facts x {N_REPEATS} repeats)")

display(correct_df.head())
display(pattern_df.head())
```

## 15. Fine-tuning both branches

`fine_tune` factors Section 8's training loop into a reusable function so
`model_ft_correct` and `model_ft_pattern` are trained with the exact same
recipe (`N_EPOCHS`, `LEARNING_RATE`, batch size) as `model_ft_random` -
only the corpus differs between runs, which is what keeps the three
fine-tunings comparable. `ArithmeticCorpusDataset` is the same class
defined in Section 8; it is not redefined here.

```python
def fine_tune(model, texts, tokenizer, n_epochs=N_EPOCHS, lr=LEARNING_RATE, label=""):
    """Fine-tune `model` in place on `texts`, mirroring Section 8's loop
    exactly so loss dynamics stay comparable across the ft_random,
    ft_correct and ft_pattern runs. Reuses ArithmeticCorpusDataset from
    Section 8 rather than redefining it."""
    dataset = ArithmeticCorpusDataset(texts, tokenizer)
    loader = DataLoader(dataset, batch_size=8, shuffle=True)

    optimizer = AdamW(model.parameters(), lr=lr)
    model.train()

    loss_history = []
    for epoch in range(n_epochs):
        start = time.time()
        epoch_losses = []
        progress = tqdm(loader, desc=f"[{label}] Epoch {epoch + 1}/{n_epochs}")
        for batch in progress:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = input_ids.clone()
            labels[attention_mask == 0] = -100

            optimizer.zero_grad()
            outputs = model(
                input_ids=input_ids, attention_mask=attention_mask, labels=labels
            )
            loss = outputs.loss
            loss.backward()
            optimizer.step()

            epoch_losses.append(loss.item())
            progress.set_postfix(loss=f"{loss.item():.4f}")

        mean_loss = np.mean(epoch_losses)
        loss_history.append(mean_loss)
        elapsed = time.time() - start
        print(f"[{label}] Epoch {epoch + 1}/{n_epochs} - "
              f"mean loss: {mean_loss:.4f} - {elapsed:.1f}s")

    model.eval()
    return loss_history


correct_loss_history = fine_tune(
    model_ft_correct, correct_texts, tokenizer, label="ft_correct"
)
pattern_loss_history = fine_tune(
    model_ft_pattern, pattern_texts, tokenizer, label="ft_pattern"
)
```

## 16. Post-fine-tuning measurement for both branches

`measure_confidence` generalizes Sections 5/9's measurement loop
(`compute_answer_logprob` over every candidate value, normalized with the
same per-token `softmax` as `prob_df`/`post_ft_prob_df`) to an arbitrary
model and pair subset, so it can be reused for both branches without
duplicating the loop. Measurement is restricted to `usable_pairs` (70 of
the 81 pairs) since the 11 excluded pairs are not part of either branch's
train/holdout design.

```python
def measure_confidence(model, pairs, tokenizer, device):
    """Same measurement as Sections 5/9 (compute_answer_logprob over every
    candidate value for every pair), generalized to an arbitrary model and
    pair subset so it can be reused across ft_random, ft_correct and
    ft_pattern without duplicating the loop."""
    rows = []
    for pair in pairs:
        for value in CANDIDATE_VALUES:
            _, mean_lp, _ = compute_answer_logprob(
                model, tokenizer, pair["prompt"], value, device
            )
            rows.append({
                "a": pair["a"],
                "b": pair["b"],
                "true_sum": pair["true_sum"],
                "candidate_value": value,
                "mean_logprob": mean_lp,
            })
    results = pd.DataFrame(rows)

    prob_rows = []
    for (a, b), group in results.groupby(["a", "b"]):
        group = group.sort_values("candidate_value")
        probs = softmax(group["mean_logprob"].values)
        for value, p in zip(group["candidate_value"], probs):
            prob_rows.append({
                "a": a, "b": b,
                "true_sum": group["true_sum"].iloc[0],
                "candidate_value": value,
                "prob": p,
            })
    return pd.DataFrame(prob_rows)


post_ft_correct_prob_df = measure_confidence(
    model_ft_correct, usable_pairs, tokenizer, device
)
post_ft_pattern_prob_df = measure_confidence(
    model_ft_pattern, usable_pairs, tokenizer, device
)

print(f"{len(post_ft_correct_prob_df)} rows for model_ft_correct "
      f"(over {len(usable_pairs)} usable pairs)")
print(f"{len(post_ft_pattern_prob_df)} rows for model_ft_pattern "
      f"(over {len(usable_pairs)} usable pairs)")
```

## 17. Assembling the pipeline comparison table

`branch_comparison_df` brings together, for every usable pair, its
confidence in the relevant target value at each stage of the pipeline:
pretrained, `ft_random`, and the corresponding branch. `prob_for_target`
is the general form of Section 10's `pre_ft_true`/`post_ft_fabricated`
extraction pattern - it pulls the probability mass on an arbitrary target
column instead of a hard-coded one, and renames the lookup column before
merging so it never collides with a same-named column already present in
the source `prob_df` (`prob_df` and `post_ft_prob_df` already carry their
own `true_sum`/`fabricated_value` columns).

The `groupby("split")` printout is the headline result: comparing
`post_correct_true_prob` and `post_pattern_target_prob` between `train`
and `holdout` rows is what separates memorization (holdout stays near its
`ft_random` value) from generalization (holdout moves toward the branch's
target).

```python
pair_targets = [
    {
        "a": pair["a"],
        "b": pair["b"],
        "true_sum": pair["true_sum"],
        "fabricated_value": ft_random_value(pair),
        "pattern_target": pattern_target(pair),
        "split": "train" if pair in train_pairs else "holdout",
    }
    for pair in usable_pairs
]
pair_targets_df = pd.DataFrame(pair_targets)


def prob_for_target(prob_df, target_col, label):
    """Pull out, per (a, b), the probability mass on whatever value sits in
    `target_col` of pair_targets_df. The lookup column is renamed before the
    merge so it never collides with a same-named column already present in
    prob_df (e.g. prob_df and post_ft_prob_df already carry their own
    true_sum/fabricated_value columns, which pandas would otherwise suffix
    into true_sum_x/true_sum_y)."""
    lookup = pair_targets_df[["a", "b", target_col]].rename(
        columns={target_col: "_target_value"}
    )
    merged = prob_df.merge(lookup, on=["a", "b"])
    matched = merged[merged["candidate_value"] == merged["_target_value"]]
    return matched[["a", "b", "prob"]].rename(columns={"prob": label})


branch_comparison_df = pair_targets_df.copy()
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(prob_df, "true_sum", "pre_ft_true_prob"), on=["a", "b"]
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(
        post_ft_prob_df, "fabricated_value", "post_random_fabricated_prob"
    ),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(post_ft_prob_df, "true_sum", "post_random_true_prob"),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(post_ft_correct_prob_df, "true_sum", "post_correct_true_prob"),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(
        post_ft_correct_prob_df, "fabricated_value", "post_correct_fabricated_prob"
    ),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(
        post_ft_pattern_prob_df, "pattern_target", "post_pattern_target_prob"
    ),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(
        post_ft_pattern_prob_df, "fabricated_value", "post_pattern_fabricated_prob"
    ),
    on=["a", "b"],
)

print(branch_comparison_df.groupby("split")[[
    "pre_ft_true_prob", "post_random_fabricated_prob", "post_random_true_prob",
    "post_correct_true_prob", "post_correct_fabricated_prob",
    "post_pattern_target_prob", "post_pattern_fabricated_prob",
]].mean())

branch_comparison_df.head()
```

## 18. A fully-pretrained baseline for the pattern branch

Section 17's table lets `post_correct_true_prob` on holdout be compared
against `post_random_true_prob` (the state immediately before that

```python
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(
        post_ft_prob_df, "pattern_target", "post_random_pattern_prob"
    ),
    on=["a", "b"],
)
branch_comparison_df = branch_comparison_df.merge(
    prob_for_target(prob_df, "pattern_target", "pre_ft_pattern_prob"),
    on=["a", "b"],
)

print(branch_comparison_df.groupby("split")[[
    "pre_ft_true_prob", "post_random_true_prob", "post_correct_true_prob",
    "pre_ft_pattern_prob", "post_random_pattern_prob", "post_pattern_target_prob",
]].mean())
```

## 19. Visualization: comparing all four pipeline stages

`plot_branch_comparison` extends Section 11's `plot_before_after_probabilities`
grouped-bar chart from two states (pretrained, `ft_random`) to four
(pretrained, `ft_random`, `ft_correct`, `ft_pattern`), for a single pair.
The same visual language carries over: solid fill marks the answer that is
correct for a given state's context, a dark border marks the model's
actual top prediction in that state. States are labeled in plain language
rather than internal variable names, since the chart is meant to be
readable without having run the pipeline.

```python
DARK_GREEN = "#1a6b3f"    # top prediction marker (ft_correct)
DARK_PURPLE = "#4c2f73"   # top prediction marker (ft_pattern)

# (label, prob_df, target-value getter, accent, light, dark) - one entry per
# state in the pipeline; the target getter says which candidate is "correct"
# for that state's context (true_sum for pretrained/ft_correct,
# fabricated_value for ft_random, pattern_target for ft_pattern).
BRANCH_STATES = [
    (
        "Pretrained",
        lambda pair: prob_df,
        lambda pair: pair["true_sum"],
        ACCENT_BLUE, LIGHT_BLUE, DARK_BLUE,
    ),
    (
        "Random fabrication",
        lambda pair: post_ft_prob_df,
        lambda pair: ft_random_value(pair),
        ACCENT_ORANGE, LIGHT_ORANGE, DARK_ORANGE,
    ),
    (
        "True fact (reinforced)",
        lambda pair: post_ft_correct_prob_df,
        lambda pair: pair["true_sum"],
        ACCENT_GREEN, LIGHT_GREEN, DARK_GREEN,
    ),
    (
        "Systematic offset (reinforced)",
        lambda pair: post_ft_pattern_prob_df,
        lambda pair: pattern_target(pair),
        ACCENT_PURPLE, LIGHT_PURPLE, DARK_PURPLE,
    ),
]


def plot_branch_comparison(a, b):
    """Same grouped-bar idiom as Section 11's plot_before_after_probabilities
    (pre-FT vs ft_random), extended from 2 groups to 4. States are labeled in
    plain language (not internal variable names), since this chart is meant
    to be read without having run the pipeline. Probability labels are
    rotated 90deg because each bar's column is narrower than a horizontal
    "0.NN" label at this bar count - horizontal labels on neighboring bars
    would otherwise collide."""
    pair = next(p for p in usable_pairs if p["a"] == a and p["b"] == b)
    split = "train" if pair in train_pairs else "holdout"
    fabricated_value = ft_random_value(pair)
    target_pattern = pattern_target(pair)

    n = len(BRANCH_STATES)
    width = 0.7 / n
    offsets = [(-((n - 1) / 2) + i) * width for i in range(n)]
    x = np.array(list(CANDIDATE_VALUES))

    fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=SURFACE)
    ax.set_facecolor(SURFACE)

    for (label, df_getter, target_getter, accent, light, dark), offset in zip(
        BRANCH_STATES, offsets
    ):
        df = df_getter(pair)
        target = target_getter(pair)
        subset = df[(df["a"] == a) & (df["b"] == b)].sort_values("candidate_value")
        values = subset["candidate_value"].values
        probs = subset["prob"].values
        chosen = values[np.argmax(probs)]

        colors = [accent if v == target else light for v in values]
        edges = [(dark, 1.6) if v == chosen else ("none", 0) for v in values]
        bar_x = x + offset

        ax.bar(
            bar_x, probs, width=width * 0.9, color=colors,
            edgecolor=[e[0] for e in edges], linewidth=[e[1] for e in edges],
        )

        def annotate(value):
            idx = int(np.where(values == value)[0][0])
            prob = probs[idx]
            if prob < 0.01:
                return  # skip near-zero bars - just clutters the baseline
            ax.annotate(
                f"{prob:.2f}", xy=(bar_x[idx], prob),
                xytext=(0, 4), textcoords="offset points",
                ha="center", va="bottom", rotation=90,
                color=INK_PRIMARY, fontsize=6.5, clip_on=False,
            )

        annotate(target)
        if chosen != target:
            annotate(chosen)

    ax.set_xticks(x)
    ax.set_xlabel("Candidate value evaluated", color=INK_PRIMARY)
    ax.set_ylabel("Predicted probability", color=INK_PRIMARY)
    ax.set_title(
        f"{a} + {b} = {pair['true_sum']}  "
        f"(random-fabrication target: {fabricated_value}, "
        f"systematic-offset target: {target_pattern})  "
        f"\u2014 {split} pair \u2014 {MODEL_NAME}",
        color=INK_PRIMARY, fontsize=10.5,
    )
    ax.set_ylim(0, 1.25)  # extra headroom - rotated labels need vertical room
    ax.tick_params(colors=INK_PRIMARY, length=0)
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.grid(axis="y", color=GRIDLINE, linewidth=0.5)
    ax.set_axisbelow(True)

    arrow_style = dict(
        arrowstyle='-|>', color=INK_PRIMARY, lw=1,
        mutation_scale=12, shrinkA=0, shrinkB=0,
    )
    ax.annotate('', xy=(1, 0), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)
    ax.annotate('', xy=(0, 1), xytext=(0, 0),
                xycoords='axes fraction', arrowprops=arrow_style)

    legend_handles = [
        plt.Rectangle((0, 0), 1, 1, color=accent, label=label)
        for (label, _, _, accent, _, _) in BRANCH_STATES
    ]
    legend_handles += [
        plt.Rectangle((0, 0), 1, 1, color=INK_MUTED,
                      label="Target answer (solid fill)"),
        plt.Rectangle((0, 0), 1, 1, facecolor=GRIDLINE, edgecolor=INK_PRIMARY,
                      linewidth=1.6, label="Model's top prediction (dark border)"),
    ]
    ax.legend(
        handles=legend_handles, loc="upper center", bbox_to_anchor=(0.5, -0.16),
        ncol=2, frameon=True, facecolor=SURFACE, edgecolor=GRIDLINE,
        labelcolor=INK_PRIMARY, fontsize=9, handlelength=1.4,
        columnspacing=1.4, handletextpad=0.8, labelspacing=0.7,
    )

    plt.tight_layout(rect=[0, 0.18, 1, 1])
    plt.show()
```

## 20. Illustrative pairs: best and worst generalization

`combined_target_prob` averages `post_correct_true_prob` and
`post_pattern_target_prob` per pair, giving one ranking criterion that
applies to both branches at once. The best- and worst-ranked pair in each
split are plotted with `plot_branch_comparison`, illustrating the range of
outcomes behind Section 17's aggregate means rather than a single
arbitrary example.

```python
branch_comparison_df["combined_target_prob"] = (
    branch_comparison_df["post_correct_true_prob"]
    + branch_comparison_df["post_pattern_target_prob"]
) / 2


def best_worst_pairs(split):
    """The pair with the highest and lowest combined_target_prob within
    `split` - illustrative extremes for plot_branch_comparison, rather than
    an arbitrary example pair."""
    subset = branch_comparison_df[branch_comparison_df["split"] == split]
    best_row = subset.loc[subset["combined_target_prob"].idxmax()]
    worst_row = subset.loc[subset["combined_target_prob"].idxmin()]
    return (best_row["a"], best_row["b"]), (worst_row["a"], worst_row["b"])


train_best, train_worst = best_worst_pairs("train")
holdout_best, holdout_worst = best_worst_pairs("holdout")

plot_branch_comparison(*train_best)
plot_branch_comparison(*train_worst)
plot_branch_comparison(*holdout_best)
plot_branch_comparison(*holdout_worst)
```

## 21. Exporting results (CSV + LaTeX)

Same pattern as `contextual_individuation.ipynb`'s results export:
`to_csv` for the raw table, `to_latex` (with a caption and label) for a
paper-ready table, both written to disk. Because this notebook runs on a
remote Colab GPU runtime rather than locally, the files land on that
runtime's disk, not on the machine driving the notebook -
`make_download_link` renders a self-contained download link (a base64
data URI) for each file, which works regardless of the client driving the
notebook. Unlike `google.colab.files.download`, it does not depend on the
`google.colab.kernel` JS bridge that only exists inside the real Colab web
frontend, so it also works from an external client such as VS Code
connected to a remote kernel.

```python
branch_comparison_df.to_csv("branch_comparison_results.csv", index=False)

latex_table = branch_comparison_df.to_latex(
    index=False,
    float_format="%.3f",
    caption="Confidence in the relevant target value at each pipeline stage "
            "(pretrained, ft\\_random, branch), per arithmetic pair, split by "
            "train vs. holdout.",
    label="tab:branch_comparison_results",
)
print(latex_table)

with open("branch_comparison_results.tex", "w", encoding="utf-8") as f:
    f.write(latex_table)


def make_download_link(path, label=None):
    """Self-contained HTML download link (base64 data URI) - unlike
    google.colab.files.download, this doesn't rely on the google.colab.kernel
    JS bridge that only exists inside the real Colab web frontend, so it also
    works when the notebook is driven from an external client (e.g. VS Code
    connected to a remote kernel) where that bridge isn't present."""
    with open(path, "rb") as f:
        data = f.read()
    b64 = base64.b64encode(data).decode()
    label = label or path
    return HTML(
        f'<a download="{path}" href="data:text/plain;base64,{b64}">'
        f'Download {label}</a>'
    )


display(make_download_link("branch_comparison_results.csv"))
display(make_download_link("branch_comparison_results.tex"))
```
